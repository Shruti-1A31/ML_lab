```python
#In this practical does not mention to implementation but this file for learning purpose example.

# Program to demonstrate various Python libraries

# 1. NumPy
import numpy as np

# Creating an array
arr = np.array([10, 20, 30, 40, 50])

print("NumPy Array:", arr)
print("Mean:", np.mean(arr))
print("Maximum:", np.max(arr))


# 2. Pandas
import pandas as pd

# Creating a DataFrame
data = {
    "Name": ["Amit", "Rahul", "Priya"],
    "Marks": [85, 90, 88]
}

df = pd.DataFrame(data)

print("\nPandas DataFrame:")
print(df)
print("Average Marks:", df["Marks"].mean())


# 3. Scikit-learn
from sklearn.linear_model import LinearRegression

# Sample data
X = np.array([[1], [2], [3], [4], [5]])
y = np.array([2, 4, 6, 8, 10])

# Create and train model
model = LinearRegression()
model.fit(X, y)

# Prediction
prediction = model.predict([[6]])

print("\nScikit-learn Prediction for 6:", prediction[0])


# 4. TensorFlow
import tensorflow as tf

# Create a constant tensor
tensor = tf.constant([[1, 2], [3, 4]])

print("\nTensorFlow Tensor:")
print(tensor)

print("Tensor Addition:")
print(tensor + 2)
```
